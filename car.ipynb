```python
import cv2 as cv
import glob
import numpy as np
import glob
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from keras.models import Sequential
from keras.layers import Dense,Flatten
from sklearn.utils import shuffle
```

```python
adress=r"C:\Users\user\Desktop\c"
```

```python
data=[]
labels=[]
```

```python
for folder in glob.glob(adress+"\\*"):
    for img_p in glob.glob(folder+"\\*"):
        img=cv.imread(img_p)
        if img is None:
            continue
        if len(img.shape)==2 and img.shape[2]==3:
            img=cv.cvtColor(img,cv.COLOR_GRAY2BGR)
        img=cv.resize(img,(64,64))
        data.append(img)
        labels.append(folder.split("\\")[-1])
```

```python
data=np.array(data,dtype="float32")/255.0
```

```python
le=LabelEncoder()
```

```python
labels=le.fit_transform(labels)
```

```python
x_train,x_test,y_train,y_test=train_test_split(data,labels,test_size=0.020,random_state=42)
```

```python
print(x_train.shape)
```

```python
print(y_train.shape)
```

```python
x_train,y_train=shuffle(x_train,y_train)
```

```python
model=Sequential([
    Flatten(input_shape=(64,64,3)),
    Dense(256,activation="relu"),
    Dense(1,activation="sigmoid")
])
```

```python
model.compile(
    optimizer="adam",loss='binary_crossentropy',metrics=['accuracy']
)
```

```python
model.fit(x_train,y_train,epochs=50,batch_size=8,validation_data=(x_test,y_test))
```

```python
model.save("modelcar.keras")
```
