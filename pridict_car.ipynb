```python
from keras.models import load_model
import cv2 as cv
import glob
import numpy as np
```

```python
model=load_model(r"c:\Users\user\Desktop\carrrrhoosh\modelcar.keras")
```

```python
adress=r"C:\Users\user\Desktop\c"
```

```python
for folder in glob.glob(adress+"\\*"):
    for img_p in glob.glob(folder+"\\*"):
        img=cv.imread(img_p)
        if img is None:
            continue
        if len(img.shape)==2 and img.shape[2]==3:
            img=cv.cvtColor(img,cv.COLOR_GRAY2BGR)
        img=cv.resize(img,(64,64))
        img=img/255.0
        img=np.expand_dims(img,axis=0)
        out=model.predict(img)
        if out[0][0]>0.5:
            print(img_p,"nocar")
        else:
            print(img_p,"car")
```
